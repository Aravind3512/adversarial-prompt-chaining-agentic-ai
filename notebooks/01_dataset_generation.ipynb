# Dataset Generation

Generate the synthetic clean, subtle-split, and strong-split datasets used by the benchmark.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from prompt_chaining_benchmark.data import generate_dataset, save_jsonl
```

```python
ratios = [0.05, 0.10, 0.20, 0.30]
datasets = {ratio: generate_dataset(ratio, total_tasks=100, seed=6730) for ratio in ratios}

for ratio, tasks in datasets.items():
    poisoned = sum(task.is_poisoned for task in tasks)
    strong = sum(task.variant == 'strong' for task in tasks)
    subtle = sum(task.variant == 'subtle' for task in tasks)
    clean = sum(task.variant == 'clean' for task in tasks)
    print(f'ratio={ratio:.2f} total={len(tasks)} poisoned={poisoned} clean={clean} strong={strong} subtle={subtle}')
```

```python
output_dir = ROOT / 'data' / 'generated'
output_dir.mkdir(parents=True, exist_ok=True)
for ratio, tasks in datasets.items():
    save_jsonl(tasks, output_dir / f'tasks_r{int(ratio * 100):02d}.jsonl')
print(f'Wrote datasets to {output_dir}')
```
