# Analyze Results

Load benchmark metrics and generate figures for ASR, clean safety, overall failure, and propagation.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from prompt_chaining_benchmark.runner import DEFAULT_RATIOS, run_experiment, save_outputs
from prompt_chaining_benchmark.plotting import read_metrics, plot_asr, plot_safety, plot_propagation
```

```python
fig_dir = ROOT / 'results' / 'figures'
paths = [plot_asr(rows, fig_dir), *plot_safety(rows, fig_dir), plot_propagation(rows, fig_dir)]
for path in paths:
    print(path)
```
